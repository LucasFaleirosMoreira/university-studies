```python
# Função para calcular a distância euclidiana
def distancia_euclidiana(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)

# Lista para armazenar as distâncias
distancias = []

# Iterando pelas linhas dos DataFrames
for _, amostra_treino in df_treino.iterrows():
    for _, amostra_teste in df_teste.iterrows():
        # Calculando a distância entre 'height' e 'weight'
        distancia = distancia_euclidiana([amostra_treino['height'], amostra_treino['weight']],
                                        [amostra_teste['height'], amostra_teste['weight']])
        distancias.append((distancia, amostra_treino[['height', 'weight']], amostra_teste[['height', 'weight']]))

# Exibindo as distâncias e as amostras
for dist in distancias:
    print(f"Distância: {dist[0]:.2f}, Amostra Treinamento: {dist[1].values}, Amostra Teste: {dist[2].values}")
```
