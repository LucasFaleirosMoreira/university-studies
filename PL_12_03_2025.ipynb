```python
T = [54, 83, 15, 71, 77, 36, 53, 38, 27, 87, 76, 91, 14, 29, 12, 77, 32, 87,
     68, 94, 79, 3, 11, 99, 56, 70, 99, 60, 5, 56, 3, 61, 73, 75, 47, 14, 21,
     86, 5, 77, 16, 89, 19] # tempos de realização das tarefas

n = 6
m = len(T)
```

```python
import gurobipy as gp

#Criando modelo
modelo = gp.Model()

#Variaveis de decisão
x = modelo.addVars(n, m, vtype = gp.GRB.BINARY)
Z = modelo.addVar()
#Função objetivo
modelo.setObjective(Z)

#Restrições
r1 = modelo.addConstrs(
    sum(x[i,j] for i in range(n)) == 1
    for j in range(m)
)

r2 = modelo.addConstrs(
    sum(T[j] * x[i,j] for j in range(m))  <= Z
    for i in range(n)
)


#Resolvendo o modelo
modelo.setParam("Outputflag", 0)
modelo.optimize()

#Apresentando a solução
if modelo.Status == 2:
  print("Valor ótimo = {}".format(modelo.objVal))
  for i in range(n):
    for j in range(m):
      print("x[{}, {}] = {} = {}".format(i+1, j+1, x[i,j].x, T[j]))
else:
  print("Não foi possivel encontrar solução ótima!!")
```
