```python
# Dados de entrada
E = [1100, 1800] # disponibilidade das fábricas
D = [960, 510, 895] # demanda dos clientes
C = [[3.7, 4.3, 6.1],
     [9.8, 6.9, 2.1]] # custos de envio
n = len(E) # número de fábricas
m = len(D) # número de clientes no
```

```python
!pip install gurobipy
import gurobipy as gp

#Criando modelo
modelo = gp.Model()

#Variáveis de decisão
x = modelo.addVars(n, m, vtype = gp.GRB.INTEGER)

#Função objetivo
modelo.setObjective(sum(C[i][j] * x[i,j] for i in range(n) for j in range(m)))

#Restrições
r1 = modelo.addConstrs(
    sum(x[i,j] for i in range(n)) == D[j]
    for j in range(m)
)

r2 = modelo.addConstrs(
    sum(x[i,j] for j in range(m)) <= E[i]
    for i in range(n)
)

#Resolvendo o modelo
modelo.setParam("OutputFlag", 0)
modelo.optimize()

#Apresentando a solução
if modelo.Status == 2:
  print("Valor ótimo = {}".format(modelo.objVal))
  for i in range(n):
    for j in range(m):
      print("x[{}, {}] = {}".format(i+1, j+1, x[i,j].x))
else:
  print("Não foi possivel encontrar solução ótima!!")
```
