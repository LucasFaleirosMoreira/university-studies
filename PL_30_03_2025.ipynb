```python
!pip install gurobipy
import gurobipy as gp


# Dados
n = 2  # Número de fábricas
m = 4  # Número de produtos
o = 3  # Número de clientes

e = [[1200, 1100, 1000, 2000], [1500, 1400, 700, 1500]]  # Capacidade (fábricas x produtos)
d = [[600, 960, 550, 1100], [855, 380, 700, 970], [720, 1120, 250, 1360]]  # Demanda (clientes x produtos)
c = [[[0.30, 1.62, 3.20, 2.00], [0.50, 2.00, 4.00, 3.60], [0.42, 1.90, 3.85, 3.00]],
     [[0.55, 2.00, 3.00, 1.90], [0.21, 1.60, 3.00, 2.80], [0.32, 1.60, 3.00, 2.80]]]  # Custos (fábricas x clientes x produtos)

modelo = gp.Model()

# Variáveis - x[fábrica, cliente, produto]
x = modelo.addVars(n, o, m, vtype= gp.GRB.INTEGER, name='x')

# Função objetivo
modelo.setObjective(
    gp.quicksum(c[i][j][k] * x[i,j,k] for i in range(n) for j in range(o) for k in range(m)),
    gp.GRB.MINIMIZE
)

# Restrições - FORMATO QUE VOCÊ PEDIU PARA MANTER
r1 = modelo.addConstrs(
    (gp.quicksum(x[i,j,k] for j in range(o)) <= e[i][k]
    for i in range(n)
    for k in range(m)
))

r2 = modelo.addConstrs(
    (gp.quicksum(x[i,j,k] for i in range(n)) == d[j][k]
    for j in range(o)
    for k in range(m)
))

# Resolver
modelo.optimize()

# Resultados
if modelo.status == gp.GRB.OPTIMAL:
    print(f"Custo total ótimo: R$ {modelo.objVal:.2f}")
    for k in range(m):
        print(f"\nProduto P{k+1}:")
        for i in range(n):
            for j in range(o):
                if x[i,j,k].x > 0:
                    print(f"  F{i+1} -> C{j+1}: {int(x[i,j,k].x)} unidades")
else:
    print("Não foi possível encontrar solução ótima!")
```
