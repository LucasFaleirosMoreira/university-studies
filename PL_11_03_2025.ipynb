```python
# Dados de entrada
n = 4 # número de ingredientes
m = 4 # número de nutrientes

P = [
    [0.26, 0.05, 0.60, 0.07],
    [0.01, 0.05, 0.75, 0.00],
    [0.25, 0.26, 0.45, 0.01],
    [0.10, 0.02, 0.24, 0.01]
]

# Custos por kg de cada ingrediente
C = [5.20, 6.80, 7.10, 2.50]

# Restrições nutricionais mínimas e máximas
min_nutrientes = [0.22, 0.07, 0, 0]
max_nutrientes = [1, 1, 0.55, 0.08]
```

```python
!pip install gurobipy
import gurobipy as gp

#Criando modelo
modelo = gp.Model()

#Variáveis de decisão
x = modelo.addVars(n, vtype=gp.GRB.CONTINUOUS, name="x")

#Função objetivo
modelo.setObjective(gp.quicksum(C[j] * x[j] for j in range(n)), gp.GRB.MINIMIZE)

#Restrições
r1 = modelo.addConstr(gp.quicksum(x[j] for j in range(n)) == 1, "Total_kg")

for i in range(m):
  modelo.addConstr(gp.quicksum(P[j][i] * x[j] for j in range(n)) >= min_nutrientes[i], f"Min_Nutriente_{i}")
  modelo.addConstr(gp.quicksum(P[j][i] * x[j] for j in range(n)) <= max_nutrientes[i], f"Max_Nutriente_{i}")


#Resolvendo o modelo
modelo.setParam("OutputFlag", 0)
modelo.optimize()

#Apresentando a solução
if modelo.status == gp.GRB.OPTIMAL:
    print(f"Valor ótimo = {modelo.objVal:.2f}")
    for j in range(n):
        print(f"x[{j+1}] = {x[j].x:.4f} kg")
else:
    print("Não foi possível encontrar solução ótima!")
```
