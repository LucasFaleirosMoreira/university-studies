```python

# Dados do problema
n = 2  # número de fábricas (F1, F2)
m = 3  # número de clientes (C1, C2, C3)
o = 4  # número de produtos (P1, P2, P3, P4)

# Capacidade de envio (e_ik): n x o
e = [
    [1200, 1100, 1000, 2000],  # F1
    [1500, 1400, 700, 1500]     # F2
]

# Demanda (d_jk): m x o
d = [
    [600, 960, 550, 1100],  # C1
    [855, 380, 700, 970],   # C2
    [720, 1120, 250, 1360]  # C3
]

# Custos (c_ijk): n x m x o
c = [
    [  # F1
        [0.30, 1.62, 3.20, 2.00],  # C1
        [0.50, 2.00, 4.00, 3.60],  # C2
        [0.42, 1.90, 3.85, 3.00]   # C3
    ],
    [  # F2
        [0.55, 2.00, 3.00, 1.90],  # C1
        [0.21, 1.60, 3.00, 2.80],  # C2
        [0.32, 1.60, 3.00, 2.80]   # C3
    ]
]
```

```python
!pip install gurobipy
import gurobipy as gp

#criando modelo
modelo = gp.Model()

#variaveis de decisão
x = modelo.addVars(n, m, o, vtype = gp.GRB.INTEGER, name = 'x')

#funçao objetivo
modelo.setObjective(
    gp.quicksum(c[i][j][k] * x[i, j, k] for i in range(n) for j in range(m) for k in range(o)),
    GRB.MINIMIZE
)

#restrições
r1 = modelo.addConstrs(
    (gp.quicksum(x[i,j,k] for j in range(m)) <= e[i][k]
    for i in range(n)
    for k in range(o)
))

r2 = modelo.addConstrs(
    (gp.quicksum(x[i,j,k] for i in range(n)) == d[j][k]
    for j in range(m)
    for k in range(o)
))





#Resolvendo o modelo
modelo.setParam("OutputFlag", 0)
modelo.optimize()

#Apresentando a solução
if modelo.status == 2:
    print(f"Custo total ótimo: R$ {modelo.objVal:.2f}")
    for k in range(o):
        print(f"\nProduto {k+1}:")
        for i in range(n):
            for j in range(m):
                if x[i,j,k].x > 0:
                    print(f"  Fabrica {i+1} para Cliente {j+1}: {int(x[i,j,k].x)} unidades")
else:
    print("Não foi possível encontrar solução ótima!")
```
