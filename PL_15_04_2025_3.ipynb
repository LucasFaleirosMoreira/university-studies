```python
# Importando biblioteca Numpy
import numpy as np
```

```python
# Dados de entrada ligas metálicas
A = np.matrix([[1, 1, 1, 0, 0],
               [3, 1, 0, 1, 0],
               [-1, 3, 0, 0, 1]])

b = np.matrix([[5],
               [12],
               [7]])

c = np.matrix([[-2],
               [-4],
               [0],
               [0],
               [0]])

base = [2, 3, 4]
```

```python
# Matriz B
B = np.matrix(A[:,base])
print("B =")
print(B)

print()

# Solução básica
xB = np.linalg.inv(B) * b
print("xB =")
print(xB)
```

```python
# Vetor pT
pT = c[base].T * np.linalg.inv(B)
print("pT = ", pT)

# Custos relativos
s = c.T - pT * A
print("s = ", s)
```
